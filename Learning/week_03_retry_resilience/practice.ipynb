## Closure

```python
def outer():
    message = 'Hi'
    def inner():       
        print(message)
    return inner
```

```python
my_func = outer()
my_func
```

```python
def outer(x):
    def inner(y):
        return x+y
    return inner
```

```python
outer1 = outer(5)
print(outer1.__name__)
print(outer1.__closure__[0].cell_contents)
print(outer1(2))
```

## Decorators

```python

def decorator_function(original_function):
    def wrapper_function(*args, **kwargs):
       print("Wrapper code executed")
       print(f"Function name is {original_function.__name__}")
       return original_function(*args, **kwargs)
    return wrapper_function

@decorator_function
def display():
    print("This is display data")

@decorator_function
def display_data(name, age):
    print (f"Name is {name}, Age is {age}")

# call1 = decorator_function(display)
# call1()

display()
display_data('Gowtham',26)
```

```python
class decorator_class:

    def __init__(self,original_funciton):
        self.original_function = original_funciton

    def __call__(self, *args, **kwargs):
        print(f"__call__ method executed this {self.original_function.__name__}")
        return self.original_function(*args,**kwargs)


@decorator_class
def display_data(name, age):
    print (f"Name is {name}, Age is {age}")

display_data('Gowtham',26)
```

```python
def my_logger(original_function):

    import logging, sys
    logging.basicConfig(
        level=logging.INFO,
        format="%(asctime)s | %(name)s | %(levelname)-8s | %(filename)s:%(lineno)d | %(message)s",
        handlers=[
            logging.StreamHandler(sys.stdout),
            logging.FileHandler("{}.log".format(original_function.__name__), mode="a"),
        ]
    )

    def wrapper(*args, **kwargs):
        logging.info(f"Ran with args: {args}, ran with kwargs:{kwargs}")
        return original_function(*args, **kwargs)
    return wrapper

@my_logger
def display_data(name, age):
    print (f"Name is {name}, Age is {age}")

display_data(name='Gowtham',age=24)
```

```python
def my_timer(original_function):
    import time
    from functools import wraps

    @wraps(original_function)
    def wrapper(*args, **kwargs):
        t1 = time.time()
        result = original_function(*args, **kwargs)
        t2 = time.time() - t1
        print(f"{original_function.__name__} function ran in {t2} sec")
        return result
    return wrapper

@my_logger
@my_timer
def display_data(name, age):
    import time
    time.sleep(1)
    print (f"Name is {name}, Age is {age}")

display_data(name='Gowtham',age=24)
```
